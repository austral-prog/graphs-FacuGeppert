# Visualización 1 — Líneas: Precio del Bitcoin por año (2015–2024)

**Qué muestra:** La evolución del precio de cierre anual de Bitcoin en dólares, desde sus primeros años hasta convertirse en uno de los activos más conocidos del mundo. Los datos tienen un orden cronológico claro. La línea conecta esa historia y permite ver los picos, caídas y explosiones de valor que vivió el Bitcoin.

```python
años_btc = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
precio_btc = [430, 963, 13850, 3742, 7240, 28990, 46306, 16547, 43599, 98000]

plt.figure(figsize=(11, 5))

plt.plot(años_btc, precio_btc, color='#F7931A', linewidth=2.5, marker='o', markersize=7)
plt.fill_between(años_btc, precio_btc, alpha=0.1, color='#F7931A')

for x, y in zip(años_btc, precio_btc):
    plt.text(x, y + 1800, f'${y:,}', ha='center', fontsize=8, color='#555')

plt.title('Precio de Cierre Anual del Bitcoin (USD)', fontsize=15, fontweight='bold')
plt.xlabel('Año')
plt.ylabel('Precio en dólares (USD)')
plt.xticks(años_btc)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('imgs/line_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Visualización 2 — Barras: Goles de Messi por club en toda su carrera

**Qué muestra:** La cantidad total de goles que marcó Lionel Messi en cada club: Barcelona, PSG, Inter Miami y Selección Argentina. Datos oficiales al 2025. Es una comparación entre categorías. La barra hace inmediatamente visible cuál fue su club más productivo y cuánto se destacó el Barcelona por sobre los demás.

```python
clubes = ['Barcelona', 'Selección\nArgentina', 'Inter Miami', 'PSG']
goles  = [672, 116, 86, 32]
colores = ['#A50044', '#75AADB', '#F7B5CD', '#003070']

plt.figure(figsize=(9, 5))

barras = plt.bar(clubes, goles, color=colores, edgecolor='white', linewidth=0.8, width=0.5)

for b in barras:
    plt.text(b.get_x() + b.get_width()/2, b.get_height() + 8,
             str(int(b.get_height())), ha='center', fontsize=13, fontweight='bold')

plt.title('Goles de Messi por Club en toda su Carrera', fontsize=15, fontweight='bold')
plt.xlabel('Club / Selección')
plt.ylabel('Goles')
plt.ylim(0, 760)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('imgs/bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Visualización 3 — Histograma: Distribución de goles de Messi por temporada

**Qué muestra:** En qué rango de goles terminó Messi más frecuentemente en sus temporadas con el Barcelona. Datos de sus 17 temporadas en el club (2004–2021). El histograma agrupa los datos en rangos y muestra con qué frecuencia Messi terminó en cada zona de goles por temporada. Permite ver en qué rango era más consistente.

```python
goles_por_temporada = np.array([
    1, 10, 16, 23, 38, 47, 53, 73, 60, 41, 58, 54, 45, 51, 31, 38, 30
])

plt.figure(figsize=(9, 5))

plt.hist(goles_por_temporada, bins=8, color='#A50044', edgecolor='black', linewidth=0.8)

promedio = np.mean(goles_por_temporada)
plt.axvline(promedio, color='#F7931A', linewidth=2.5,
            linestyle='--', label=f'Promedio: {promedio:.1f} goles')

plt.title('Distribución de Goles de Messi por Temporada en el Barcelona', fontsize=14, fontweight='bold')
plt.xlabel('Goles en la temporada')
plt.ylabel('Cantidad de temporadas')
plt.legend(fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('imgs/histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Visualización 4 — Dispersión: Precio del Oro vs. Bitcoin (2015–2024)

**Qué muestra:** Cada punto representa un año. El eje X es el precio del oro (en USD por onza) y el eje Y es el precio del Bitcoin. Se puede ver cómo ambos activos crecieron, pero a velocidades muy distintas. Este gráfico permite explorar si existe una relación entre dos activos considerados 'reserva de valor'. Se ve claramente que el Bitcoin creció mucho más de forma explosiva.

```python
precio_oro  = np.array([1060, 1145, 1305, 1282, 1478, 1887, 1829, 1800, 1940, 2390])
precio_btc2 = np.array([430, 963, 13850, 3742, 7240, 28990, 46306, 16547, 43599, 98000])
años_label  = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

plt.figure(figsize=(10, 6))

scatter = plt.scatter(precio_oro, precio_btc2,
                      c=años_label, cmap='plasma',
                      s=120, zorder=3, edgecolors='black')

for i, año in enumerate(años_label):
    plt.text(precio_oro[i] + 10, precio_btc2[i] + 800, str(año), fontsize=9, color='#333')

plt.colorbar(scatter, label='Año')

plt.title('Precio del Oro vs. Bitcoin por Año (2015–2024)', fontsize=14, fontweight='bold')
plt.xlabel('Precio del Oro (USD/onza)')
plt.ylabel('Precio del Bitcoin (USD)')
plt.grid(linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('imgs/scatter_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Visualización 5 — Barras Horizontales: Las 8 ciudades más pobladas del mundo (2024)

**Qué muestra:** El ranking de las 8 ciudades más habitadas del planeta según datos de la ONU 2024, con sus poblaciones en millones de habitantes. Las etiquetas de las ciudades son largas y se leen mejor en horizontal. Además, el formato de ranking queda muy natural en este tipo de gráfico: la más poblada arriba y el resto hacia abajo.

```python
ciudades   = ['Tokio', 'Delhi', 'Shanghái', 'Dhaka', 'São Paulo', 'El Cairo', 'Ciudad de México', 'Mumbai']
poblacion  = [37.1, 33.8, 29.9, 23.9, 22.8, 22.6, 22.5, 21.7]
colores    = ['#E63946', '#457B9D', '#E9C46A', '#264653', '#2A9D8F', '#F4A261', '#A8DADC', '#6D6875']

ciudades_inv  = ciudades[::-1]
poblacion_inv = poblacion[::-1]
colores_inv   = colores[::-1]

plt.figure(figsize=(10, 6))

barras = plt.barh(ciudades_inv, poblacion_inv, color=colores_inv, edgecolor='black', height=0.6)

for b in barras:
    ancho = b.get_width()
    plt.text(ancho + 0.2, b.get_y() + b.get_height()/2,
             f'{ancho} M', va='center', fontsize=10, fontweight='bold')

plt.title('Las 8 Ciudades más Pobladas del Mundo (2024)', fontsize=14, fontweight='bold')
plt.xlabel('Población (millones de habitantes)')
plt.xlim(0, 41)
plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('imgs/custom_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
